## 2. The question: decision, action, cost of a wrong call

# Research Question

**Research Question**

Can machine learning help prioritize webpages that should be reviewed for content refresh?

**Unit of Analysis**

One webpage.

**Decision**

Should this webpage be prioritized for review?

**Action**

The SEO or content team reviews and updates the recommended pages.

**Cost of a Wrong Recommendation**

- False Positive: A healthy page is reviewed unnecessarily, increasing workload.
- False Negative: A declining page is missed, which may result in continued traffic loss.

Machine learning can help identify useful patterns in historical search and engagement data that are difficult to detect manually.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Total Pages:", len(df))
print("Total Columns:", len(df.columns))

declining = (df["trend_direction"] == "down").mean()
print("Percentage of Declining Pages:", round(declining * 100, 2), "%")

corr = df["search_volume"].corr(df["impressions_90d"])
print("Correlation between Search Volume and Impressions:", round(corr, 3))
```
